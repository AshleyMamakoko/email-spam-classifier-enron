```python
# Decision Tree Spam Classifier using ENRON Dataset

import numpy as np
import pandas as pd
from datasets import load_dataset

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Load Real ENRON Spam Dataset from HuggingFace
ds = load_dataset("bvk/ENRON-spam")

# Convert to pandas DataFrame
df = ds["train"].to_pandas()

# FIX: Flatten the email list
df["email"] = df["email"].apply(lambda x: x[0] if isinstance(x, list) else x)

# Keep needed columns
df = df[["email", "label"]]

print("Dataset preview after fix:")
print(df.head())
```

```python
# Train/Test Split

X_train, X_test, y_train, y_test = train_test_split(
    df["email"],
    df["label"],
    test_size=0.2,
    random_state=42,
    stratify=df["label"]
)
```

```python
# Text → TF-IDF Vectorization

vectorizer = TfidfVectorizer(
    ngram_range=(1,2),
    min_df=3,
    stop_words="english"
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```python
# Train Decision Tree Classifier

clf = DecisionTreeClassifier(
    max_depth=12,
    min_samples_split=5,
    random_state=42
)

clf.fit(X_train_tfidf, y_train)
```

```python
# Evaluation

y_pred = clf.predict(X_test_tfidf)
accuracy = accuracy_score(y_test, y_pred)

print(f"\nModel Accuracy: {accuracy:.3f}\n")

print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=["HAM", "SPAM"]))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# Inspect Decision Tree

feature_names = np.array(vectorizer.get_feature_names_out())
tree_text = export_text(clf, feature_names=list(feature_names))

print("\nDecision Tree (first 1000 chars):\n")
print(tree_text[:1000] + ("\n... (truncated)" if len(tree_text) > 1000 else ""))
```

```python
# Most Important Features

importances = clf.feature_importances_
idx = np.where(importances > 0)[0]

top_idx = idx[np.argsort(importances[idx])][-20:]  # top 20

top_features = sorted(
    zip(importances[top_idx], feature_names[top_idx]),
    reverse=True
)

print("\nTop 20 Most Important Features:")
for score, feat in top_features:
    print(f"{score:.4f}  {feat}")
```

```python
# 8) Example Predictions (Realistic)

samples = [
    "Free money!!! Claim your prize now by clicking the link",
    "Hi team, please see the attached meeting agenda for tomorrow",
    "URGENT! Your account has been suspended. Verify immediately",
    "Lunch at 1pm? Let me know if you're free"
]

X_samples = vectorizer.transform(samples)
predictions = clf.predict(X_samples)

print("\nSample Predictions:")
for text, p in zip(samples, predictions):
    print("\nEmail:", text)
    print("Predicted:", "SPAM" if p == 1 else "HAM")
```
